# Indexing, Slicing

single element extraction = indexing
multi element extraction = slicing

slicing generic formula = arr[ row_start : row_stop , col_start : col_stop ]

## Common Patterns
## What you want to extract   

1. A single specific element arr[row, col]  arr12[3, 2] -> 14
2. An entire single row arr[row_index, :]arr12[2, :] -> [8, 9, 10, 11]
3. An entire single column  arr[:, col_index]arr12[:, 1] -> [1, 5, 9, 13, 17, 21]
4. First 2 rows, all columns arr[-2:, -2:] -> [[18, 19], [22, 23]]
5. Last 2 rows, last 2 columns  arr[-2:, -2:] -> [[18, 19], [22, 23]]
6. Every alternate row/col arr[::2, ::2]	Takes step size of 2 across both dimensions
