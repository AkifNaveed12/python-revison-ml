```python
# dot product
arr3 = np.arange(6).reshape(2,3)
arr4 = np.arange(6,12).reshape(3,2)
```

```python
arr3.dot(arr4)  # row1 x col1 * row2 x col2
# here col1 and row2 must be same so the resulting matrix order will be row1 x col2
```
