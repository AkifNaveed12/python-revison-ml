## stacking() a trick to combine 2 arrays

```python
arr5=np.arange(12,18).reshape(2,3)
arr5
```

```python
arr3 = np.arange(6).reshape(2,3)
arr3
```

```python
np.hstack((arr3, arr5)) # hstack horizontally stack krta ha 
```

```python
np.vstack((arr3,arr5)) # vstack vertically stack krta ha
```

## spliting
