```python
a=np.random.randint(1,10,15).reshape(3,5)
a
```

```python
np.max(a)
```

```python
# argmax -> returns the max no index
np.argmax(a)
```

```python
# argmin -> returns the min no index
np.argmin(a)
```

```python
a[a%2==1]=-1 # replacing odd no with -1, but its changing the original array
a
```

```python
# using where when we dont need to change the original array
a=np.random.randint(1,20,6)
a
```

```python
# creating another arry for modifications and original array remains the same
np.where(a%2==1, -1, a)
```

```python
a # not changed!
```
